```python
import os

COMBINED_DIR = "./combined_dataset/train"
SELECTED_CLASSES = [
    "Tomato___Early_blight",
    "Tomato___Late_blight",
    "Tomato___healthy",
    "Tomato___Tomato_Yellow_Leaf_Curl_Virus",
    "Tomato___Leaf_Mold",
]

print("=" * 70)
print("COMBINED DATASET VERIFICATION")
print("=" * 70)

total_images = 0
for cls in SELECTED_CLASSES:
    cls_path = os.path.join(COMBINED_DIR, cls)
    
    if not os.path.exists(cls_path):
        print(f"MISSING  : {cls}")
        continue
    
    all_files   = os.listdir(cls_path)
    image_files = [f for f in all_files if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
    pv_images   = [f for f in image_files if not f.startswith('plantdoc_')]
    pd_images   = [f for f in image_files if f.startswith('plantdoc_')]
    
    total_images += len(image_files)
    print(f"\n{cls}")
    print(f"  PlantVillage images : {len(pv_images)}")
    print(f"  PlantDoc images     : {len(pd_images)}")
    print(f"  Total               : {len(image_files)}")

print("\n" + "=" * 70)
print(f"TOTAL IMAGES ACROSS ALL CLASSES: {total_images}")
print("=" * 70)

# Check for any non-image files that snuck in
print("\nChecking for non-image files...")
for cls in SELECTED_CLASSES:
    cls_path = os.path.join(COMBINED_DIR, cls)
    if not os.path.exists(cls_path):
        continue
    bad_files = [f for f in os.listdir(cls_path) 
                 if not f.lower().endswith(('.jpg', '.jpeg', '.png'))]
    if bad_files:
        print(f"  Non-image files in {cls}: {bad_files}")

print("\nDone.")
```

```python
import os, shutil

# Corrected mapping matching your exact PlantDoc folder names
PLANTDOC_TO_PLANTVILLAGE = {
    "Tomato_Early_blight_leaf"   : "Tomato___Early_blight",
    "Tomato_leaf_late_blight"    : "Tomato___Late_blight",
    "Tomato_leaf"                : "Tomato___healthy",
    "Tomato_leaf_yellow_virus"   : "Tomato___Tomato_Yellow_Leaf_Curl_Virus",
    "Tomato_mold_leaf"           : "Tomato___Leaf_Mold",
}

PLANTDOC_DIR = "./Plantdoc_dataset/train"
COMBINED_DIR = "./combined_dataset/train"

print("Adding PlantDoc images to combined dataset...")
print("=" * 60)

for plantdoc_name, standard_name in PLANTDOC_TO_PLANTVILLAGE.items():
    src = os.path.join(PLANTDOC_DIR, plantdoc_name)
    dst = os.path.join(COMBINED_DIR, standard_name)

    if not os.path.exists(src):
        print(f"NOT FOUND: '{plantdoc_name}'")
        continue

    os.makedirs(dst, exist_ok=True)

    image_files = [f for f in os.listdir(src)
                   if f.lower().endswith(('.jpg', '.jpeg', '.png'))]

    copied = 0
    for fname in image_files:
        dst_fname = f"plantdoc_{fname}"
        dst_path  = os.path.join(dst, dst_fname)

        # Skip if already copied from a previous run
        if os.path.exists(dst_path):
            continue

        shutil.copy2(os.path.join(src, fname), dst_path)
        copied += 1

    print(f"  {plantdoc_name:<40} → +{copied} images added")

print("\nDone. Running final verification...")
print("=" * 60)

SELECTED_CLASSES = [
    "Tomato___Early_blight",
    "Tomato___Late_blight",
    "Tomato___healthy",
    "Tomato___Tomato_Yellow_Leaf_Curl_Virus",
    "Tomato___Leaf_Mold",
]

total = 0
for cls in SELECTED_CLASSES:
    cls_path  = os.path.join(COMBINED_DIR, cls)
    all_imgs  = [f for f in os.listdir(cls_path)
                 if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
    pv_imgs   = [f for f in all_imgs if not f.startswith('plantdoc_')]
    pd_imgs   = [f for f in all_imgs if f.startswith('plantdoc_')]
    total    += len(all_imgs)
    print(f"{cls}")
    print(f"  PlantVillage: {len(pv_imgs):>4}  |  PlantDoc: {len(pd_imgs):>3}  |  Total: {len(all_imgs):>4}")

print(f"\nGRAND TOTAL: {total} images")
```

```python
import os, shutil

PLANTDOC_TO_PLANTVILLAGE = {
    "Tomato_Early_blight_leaf"  : "Tomato___Early_blight",
    "Tomato_leaf_late_blight"   : "Tomato___Late_blight",
    "Tomato_leaf"               : "Tomato___healthy",
    "Tomato_leaf_yellow_virus"  : "Tomato___Tomato_Yellow_Leaf_Curl_Virus",
    "Tomato_mold_leaf"          : "Tomato___Leaf_Mold",
}

PLANTDOC_TEST_DIR  = "./Plantdoc_dataset/test"
REALWORLD_TEST_DIR = "./realworld_test"   # separate folder, never mixed with training

print("PlantDoc TEST set — real-world generalization data")
print("=" * 60)

for plantdoc_name, standard_name in PLANTDOC_TO_PLANTVILLAGE.items():
    src = os.path.join(PLANTDOC_TEST_DIR, plantdoc_name)
    dst = os.path.join(REALWORLD_TEST_DIR, standard_name)

    if not os.path.exists(src):
        print(f"NOT FOUND in test: '{plantdoc_name}'")
        continue

    os.makedirs(dst, exist_ok=True)

    image_files = [f for f in os.listdir(src)
                   if f.lower().endswith(('.jpg','.jpeg','.png'))]

    for fname in image_files:
        shutil.copy2(
            os.path.join(src, fname),
            os.path.join(dst, f"plantdoc_test_{fname}")
        )

    print(f"  {plantdoc_name:<40} → {len(image_files)} test images")

print("\nReal-world test set ready at:", REALWORLD_TEST_DIR)
```

```python
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

REALWORLD_TEST_DIR = "./realworld_test"
SELECTED_CLASSES   = [
    "Tomato___Early_blight",
    "Tomato___Late_blight",
    "Tomato___healthy",
    "Tomato___Tomato_Yellow_Leaf_Curl_Virus",
    "Tomato___Leaf_Mold",
]

# Same transforms as your standard test set — no augmentation
eval_transforms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

realworld_dataset = datasets.ImageFolder(
    root=REALWORLD_TEST_DIR,
    transform=eval_transforms
)

# Force same class mapping as training set
realworld_dataset.class_to_idx = {c: i for i, c in enumerate(SELECTED_CLASSES)}
realworld_dataset.classes      = SELECTED_CLASSES
realworld_dataset.samples = [
    (p, realworld_dataset.class_to_idx[os.path.basename(os.path.dirname(p))])
    for p, _ in realworld_dataset.samples
    if os.path.basename(os.path.dirname(p)) in realworld_dataset.class_to_idx
]
realworld_dataset.targets = [s[1] for s in realworld_dataset.samples]

realworld_loader = DataLoader(
    realworld_dataset,
    batch_size=4,
    shuffle=False,
    num_workers=2
)

print(f"Real-world test images: {len(realworld_dataset)}")
for cls in SELECTED_CLASSES:
    count = sum(1 for _, l in realworld_dataset.samples 
                if l == realworld_dataset.class_to_idx[cls])
    print(f"  {cls:<50} {count} images")
```
